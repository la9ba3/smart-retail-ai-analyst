```python
import pandas as pd 
from pathlib import Path 
import matplotlib.pyplot as plt
```

```python
rfm = pd.read_csv("../data/processed/customer_rfm.csv")
```

```python
rfm.head()
```

```python
rfm.columns.tolist()
```

```python
segment_counts = rfm[" segment"].value_counts()

segment_counts
```

```python
plt.figure(figsize=(8, 5))
plt.bar(segment_counts.index, segment_counts.values)
plt.title("Customer Segments - RFM")
plt.xlabel("Segment")
plt.ylabel("Number of Customers")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
rfm.groupby(" segment")[[" recency", " frequency", " monetary"]].mean().sort_values(
    " monetary",
    ascending=False
)
```

## Conclusion

Le scoring RFM permet de transformer les données clients en segments exploitables.

Les clients `Champions` sont récents, fréquents et dépensent beaucoup.

Les clients `Loyal` achètent régulièrement.

Les clients `At risk` ont déjà acheté plusieurs fois mais ne sont plus récents.

Les clients `Lost` sont peu actifs et anciens.

Les clients `Potential` peuvent être travaillés avec des actions marketing ciblées.

Cette segmentation peut aider à personnaliser les campagnes de fidélité.
