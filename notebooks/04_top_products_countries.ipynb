```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
top_products_revenue = pd.read_csv("../data/processed/top_products_by_revenue.csv")
top_products_quantity = pd.read_csv("../data/processed/top_products_by_quantity.csv")
top_countries_revenue = pd.read_csv("../data/processed/top_countries_by_revenue.csv")
```

```python
top_products_revenue
top_products_quantity
top_countries_revenue
```

```python
plt.figure(figsize=(12, 6))
plt.barh(top_products_revenue["Description"], top_products_revenue["total_revenue"])
plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.barh(top_products_quantity["Description"], top_products_quantity["total_quantity"])
plt.title("Top 10 Products by Quantity Sold")
plt.xlabel("Quantity Sold")
plt.ylabel("Product")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.barh(top_countries_revenue["Country"], top_countries_revenue["total_revenue"])
plt.title("Top 10 Countries by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Country")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## Conclusion

Cette analyse permet d'identifier les produits et les pays qui contribuent le plus aux ventes.

Les tops produits par chiffre d'affaires montrent les articles les plus importants financièrement.

Les tops produits par quantité montrent les articles les plus vendus en volume.

Le top pays montre les marchés les plus importants pour l'entreprise.

Ces résultats peuvent aider à prioriser les produits, les stocks, les campagnes marketing et les analyses géographiques.
